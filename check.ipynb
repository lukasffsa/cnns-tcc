```python
import pandas as pd

alfafam1 = pd.DataFrame(pd.read_csv('alfafa-23-10-25/altura_alfafa_23_10_25.csv', sep=";", encoding='latin1'))
alfafam2 = pd.DataFrame(pd.read_csv('alfafa-23-11-25/Alturas_alfafa.csv', sep=";", encoding='latin1'))
df3 = pd.DataFrame(pd.read_csv('alfafa-23-11-25/Alturas_alfafa2.csv', sep=";", encoding='latin1'))
df4 = pd.DataFrame(pd.read_csv('alfafa-23-11-25/Pmassa_alfafa.csv', sep=";", encoding='latin1'))


display(1)
# display(df4)
```

```python
merge = pd.concat([alfafam1, alfafam2], ignore_index=True)
merge.insert(1, 'Cultivar', merge['ALTURA'].str[0:2])
display(merge)
```

```python
merge["PARCELA"] = merge["ALTURA"].str[:4]

# Fazer o merge
df_final = merge.merge(
    df4,
    on="PARCELA",
    how="left"
)

display(df_final)

df_final.to_csv('dataset/metadata.csv', sep=";", index=False, encoding='latin1')
```
